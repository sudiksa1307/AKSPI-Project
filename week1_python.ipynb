```python
import pandas as pd

url = "https://raw.githubusercontent.com/sudiksa1307/AKSPI-Project/refs/heads/main/heart_disease.csv"
df = pd.read_csv(url)
df.head
```

```python
df.info
```

```python
print(df.duplicated().sum(), "duplicates")
print(df.isna().sum())

df = df.drop_duplicates()
df = df.dropna()
print(df.shape)
```

```python
print(df.columns)
# Assuming you want to group by 'sex' and sum the 'num' column (likely representing heart disease cases)
# Replace 'sex' and 'num' with appropriate column names if your intent was different.
df.groupby("sex")["num"].sum().plot(kind = "bar", title = "Heart Disease Cases by Sex")
```
