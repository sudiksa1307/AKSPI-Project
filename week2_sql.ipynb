```python
import pandas as pd
import sqlite3

df = pd.read_csv("https://raw.githubusercontent.com/sudiksa1307/AKSPI-Project/refs/heads/main/heart_disease.csv")
df = df.drop_duplicates().dropna()
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
conn = sqlite3.connect(":memory:")
df.to_sql("data", conn, index=False)

def q(sql):
    return pd.read_sql(sql, conn)
```

My table has 5,000 rows.
One row represents one patient record.
Number to measure: percentage of patients with heart disease, using the num column.
Groups to compare: sex.

```python
q("""
SELECT COUNT(*) AS patient_count
FROM data
WHERE sex = 0
""")
```

```python
q("""
SELECT COUNT(*) AS patient_count
FROM data
WHERE sex = 1
""")
```

```python
q("""
SELECT sex,
       COUNT(*) AS patient_count
FROM data
GROUP BY sex
""")
```

```python
q("""
SELECT sex,
       COUNT(*) AS patient_count
FROM data
GROUP BY sex
ORDER BY patient_count DESC
LIMIT 5
""")
```

Q1: How many patient records show heart disease?
Columns needed: num
Q2: Which sex group has the highest percentage of patients with heart disease?
Columns needed: sex, num
Q3: What are the top five ages by number of heart disease cases among patients with sex = 1?
Columns needed: age, sex, num

```python
df.columns
```

```python
q("""
SELECT COUNT(*) AS heart_disease_cases
FROM data
WHERE num > 0
""")
```

```python
q("""
SELECT sex,
       COUNT(*) AS total_patients,
       SUM(CASE WHEN num > 0 THEN 1 ELSE 0 END) AS disease_cases,
       ROUND(100.0 * AVG(CASE WHEN num > 0 THEN 1 ELSE 0 END), 2)
           AS disease_percentage
FROM data
GROUP BY sex
ORDER BY disease_percentage DESC
""")
```

```python
q("""
SELECT age,
       COUNT(*) AS disease_cases
FROM data
WHERE sex = 1 AND num > 0
GROUP BY age
ORDER BY disease_cases DESC, age ASC
LIMIT 5
""")
```
